```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

np.random.seed(42)
print('Libraries imported successfully.')
```

## 1. Generate Synthetic Dataset

```python
n=500
attendance=np.clip(np.random.normal(78,12,n),35,100)
study_hours=np.clip(np.random.normal(3.5,1.5,n),0.5,9)
previous_score=np.clip(np.random.normal(68,15,n),20,100)
assignment_score=np.clip(np.random.normal(72,14,n),20,100)
participation=np.clip(np.random.normal(70,18,n),10,100)
final_score=np.clip(0.25*attendance+3.2*study_hours+0.35*previous_score+0.20*assignment_score+0.15*participation+np.random.normal(0,7,n),0,100)
academic_risk=pd.cut(final_score,bins=[-np.inf,50,70,np.inf],labels=['High Risk','Medium Risk','Low Risk'])
df=pd.DataFrame({'attendance':attendance.round(2),'study_hours':study_hours.round(2),'previous_score':previous_score.round(2),'assignment_score':assignment_score.round(2),'participation':participation.round(2),'final_score':final_score.round(2),'academic_risk':academic_risk})
df.head()
```

```python
print('Shape:',df.shape)
display(df.describe(include='all'))
print('\nMissing values:')
print(df.isnull().sum())
```

## 2. Exploratory Data Analysis

```python
plt.figure(figsize=(8,5))
sns.histplot(df['final_score'],bins=25,kde=True)
plt.title('Distribution of Final Scores')
plt.xlabel('Final Score'); plt.ylabel('Number of Students'); plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x='study_hours',y='final_score',hue='academic_risk')
plt.title('Study Hours vs Final Score'); plt.show()
```

```python
numeric_cols=['attendance','study_hours','previous_score','assignment_score','participation','final_score']
plt.figure(figsize=(9,6))
sns.heatmap(df[numeric_cols].corr(),annot=True,fmt='.2f',cmap='coolwarm')
plt.title('Correlation Matrix'); plt.show()
```

## 3. Machine Learning Model

```python
features=['attendance','study_hours','previous_score','assignment_score','participation']
X=df[features]; y=df['academic_risk']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
model=RandomForestClassifier(n_estimators=200,random_state=42)
model.fit(X_train,y_train)
pred=model.predict(X_test)
print('Accuracy:',round(accuracy_score(y_test,pred),4))
print('\nClassification Report:\n',classification_report(y_test,pred))
```

```python
cm=confusion_matrix(y_test,pred,labels=model.classes_)
plt.figure(figsize=(7,5)); sns.heatmap(cm,annot=True,fmt='d',xticklabels=model.classes_,yticklabels=model.classes_)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix'); plt.show()
```

```python
importance=pd.Series(model.feature_importances_,index=features).sort_values(ascending=False)
plt.figure(figsize=(8,5)); importance.plot(kind='bar')
plt.title('Feature Importance'); plt.ylabel('Importance'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
print(importance)
```

## 4. Example Prediction

```python
example_student=pd.DataFrame([{'attendance':82,'study_hours':4.5,'previous_score':72,'assignment_score':78,'participation':75}])
print('Illustrative predicted academic-risk category:',model.predict(example_student)[0])
```
